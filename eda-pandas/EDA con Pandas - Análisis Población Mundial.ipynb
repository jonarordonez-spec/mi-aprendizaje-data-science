```python
#Para hacer una correlacion 
#Seleccionar columnas numéricas explícitamente
df_numeric = df.select_dtypes(include=['number'])
df_numeric.corr()

# Opción 2: Usar numeric_only=True (en versiones más recientes de pandas)
df.corr(numeric_only=True)

#Lo que hace es comparar cada columna con la columna de enseguida y saber que tan correlacionadas estan una de la otra 
```

```python
# Vamos a ver la correlacion que tienen en un heatmap
sns.heatmap(df.corr(numeric_only=True), annot = True)

plt.rcParams['figure.figsize'] = (22,7)
plt.show()
```

```python
#Agrupa los continentes y saca la media de los datos numericos 

df.groupby('Continent').mean(numeric_only=True)

```

```python
df[df['Continent'].str.contains('Oceania')] #Muestra todos los paises de Oeania
```

```python
df2 = df.groupby('Continent')[df.columns[5:13]].mean(numeric_only=True)
df2_sorted = df2.sort_values(by='2022 Population', ascending=False)
df2_sorted

#Este código agrupa por continente y calcula promedios de variables poblacionales/históricas 
#y demográficas para cada continente,
#ignorando columnas de texto como nombres de países o capitales.
```

```python
df2_sorted.plot()
```

```python
df2_sorted.transpose() #Para cambiar las columnas a las filas y viceversa
```

```python
df2_sorted.transpose().plot()
```
